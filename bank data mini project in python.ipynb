# Excel File

# Total Bank Balance

```python
total_balance = df["Balance"].sum()

print("Total Bank Balance:", total_balance)
```

# Average Balance

# Minimum Balance

# Account Type Analysis

```python
account_type_count = df["Account_Type"].value_counts()

print(account_type_count)
```

# Pie Chart

```python
plt.figure(figsize=(7,7))

account_type_count.plot(
    kind="pie",
    
    autopct="%1.1f%%"
)

plt.title("Account Type Distribution")
plt.ylabel("")
plt.show()
```

# Gender Analysis

```python
gender_count = df["Gender"].value_counts()

print(gender_count)
```

# Bar Chart

```python
plt.figure(figsize=(7,5))

gender_count.plot(kind="bar" , color = "#E73F1E")

plt.title("Customers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers" )
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

# Pie Chart

# Loan Status Analysis

```python
loan_count = df["Loan_Status"].value_counts()

print(loan_count)
```

# Bar Chart

```python
plt.figure(figsize=(8,5))

loan_count.plot(kind="bar", color = "#853953")

plt.title("Loan Status Analysis")
plt.xlabel("Loan Status")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# City-Wise Customers

```python
city_count = df["City"].value_counts()

print(city_count)
```

# Top Cities Chart

```python
plt.figure(figsize=(10,5))

city_count.head(10).plot(kind="bar", color = "#F599C6")

plt.title("Top 10 Cities by Number of Customers")
plt.xlabel("City")
plt.ylabel("Number of Customers")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Age Analysis

```python
plt.figure(figsize=(8,5))

plt.hist(df["Age"], bins=10, color = "#8C56D4")

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

# Credit Score Analysis

```python
plt.figure(figsize=(8,5))

plt.hist(df["Credit_Score"], bins=10, color = "#E76F51")

plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

# Top 10 Customers By Balance

```python
top_balance = df.sort_values(
    by="Balance",
    ascending=False
).head(10)

print(top_balance[
    ["Customer_Name", "Balance", "Account_Type"]
])
```

# Chart

```python
plt.figure(figsize=(10,5))

plt.bar(
    top_balance["Customer_Name"],
    top_balance["Balance"]
, color = "#FFF449")

plt.title("Top 10 Customers by Account Balance")
plt.xlabel("Customer")
plt.ylabel("Balance")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Top 10 Customers By Transactions

```python
top_transactions = df.sort_values(
    by="Transactions",
    ascending=False
).head(10)

print(top_transactions[
    ["Customer_Name", "Transactions", "Balance"]
])
```

# Chart

```python
plt.figure(figsize=(10,5))

plt.bar(
    top_transactions["Customer_Name"],
    top_transactions["Transactions"]
, color = "#E76F51")

plt.title("Top 10 Customers by Transactions")
plt.xlabel("Customer")
plt.ylabel("Transactions")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Numpy Analysis

```python
balance_array = df["Balance"].to_numpy()

print("Mean Balance:", np.mean(balance_array))
print("Maximum Balance:", np.max(balance_array))
print("Minimum Balance:", np.min(balance_array))
print("Standard Deviation:", np.std(balance_array))
```

# High Balance Customers

```python
high_balance = df[df["Balance"] > 100000]

print("Customers with Balance > 100000:")
print(high_balance.head(10))

print("Total Customers:", len(high_balance))
```

# Good Credit Score Customers

```python
good_credit = df[df["Credit_Score"] >= 750]

print("Customers with Credit Score >= 750:")
print(good_credit.head(10))

print("Total Customers:", len(good_credit))
```

# Active Accounts 

```python
active_accounts = df[
    df["Account_Status"] == "Active"
]

print(active_accounts.head())

print("Total Active Accounts:", len(active_accounts))
```

# Account Type Vs Average Balance

```python
account_balance = df.groupby(
    "Account_Type"
)["Balance"].mean()

print(account_balance)
```

# Chart

```python
plt.figure(figsize=(8,5))

account_balance.plot(kind="bar",color = "#BC4F4F")

plt.title("Average Balance by Account Type")
plt.xlabel("Account Type")
plt.ylabel("Average Balance")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Account Type Vs Transactions

```python
account_transactions = df.groupby(
    "Account_Type"
)["Transactions"].mean()

print(account_transactions)
```

# Chart

```python
plt.figure(figsize=(8,5))

account_transactions.plot(kind="bar", color = "#E63946")

plt.title("Average Transactions by Account Type")
plt.xlabel("Account Type")
plt.ylabel("Average Transactions")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
